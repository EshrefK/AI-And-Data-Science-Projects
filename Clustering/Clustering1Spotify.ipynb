```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

features = [
    "danceability", "energy", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness",
    "valence", "tempo", "duration_min",
]

songs = df.drop_duplicates(subset="track_id")
songs = songs[songs["duration_min"] <= 10].copy()

x_scaled = StandardScaler().fit_transform(songs[features])
```

```python
from kneed import KneeLocator

inertias = []
for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(x_scaled)
    inertias.append(model.inertia_)

kl = KneeLocator(range(2, 11), inertias, curve="convex", direction="decreasing")
print("Elbow k:", kl.elbow)
plt.figure(figsize=(7, 4))
plt.plot(range(2, 11), inertias, marker="o")
plt.axvline(kl.elbow, color="red", linestyle="--")
plt.xlabel("k")
plt.ylabel("Inertia")
plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=6, random_state=42, n_init=10)
songs["cluster"] = kmeans.fit_predict(x_scaled)

songs.groupby("cluster")[features].mean().round(2)
```

```python
profile = songs.groupby("cluster")[features].mean()
profile_z = (profile - profile.mean()) / profile.std()

plt.figure(figsize=(10, 5))
sns.heatmap(profile_z, cmap="coolwarm", center=0, annot=True, fmt=".2f")
plt.title("Cluster profile")
plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import PCA
import numpy as np

pca = PCA(n_components=2, random_state=42)
xy = pca.fit_transform(x_scaled)

rng = np.random.RandomState(42)
idx = rng.choice(len(songs), size=8000, replace=False)

plt.figure(figsize=(8, 6))
sns.scatterplot(x=xy[idx, 0], y=xy[idx, 1], hue=songs["cluster"].to_numpy()[idx], palette="husl", s=8, alpha=0.5)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

labels = songs["cluster"].to_numpy()
print(silhouette_score(x_scaled[idx], labels[idx]))
```

# Conclusion

I cleaned the Spotify data, dropped missing and duplicate rows, and turned duration into minutes. For clustering I kept the audio features, removed duplicate tracks and songs longer than 10 minutes, then scaled the columns. KneeLocator picked 6 clusters. The groups split mostly by energy, acousticness, instrumentalness and speechiness: one group is spoken and live, one is loud and fast, one is instrumental, one is quiet acoustic instrumental, one is acoustic with vocals, and one is danceable and positive. A PCA scatter shows these groups in two dimensions, and the silhouette score on an 8000 track sample measures how separated they are.
